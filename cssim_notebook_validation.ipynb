## Sistemas de primeira ordem

<center>
<img src="img/first_order.png" alt="Sistema de primeira ordem" width="60%">
</center>

Considerando o sistema de primeira ordem mostrado acima, a relação entrada-saída é dada por:

\begin{equation}
\frac{C(s)}{R(s)} = \frac{1}{T s + 1}.
\tag{1.1}
\end{equation}

A seguir analisaremos as respostas do sistema a entradas com funções degrau unitário e rampa unitária, considerando condições iniciais nulas.


### Resposta ao degrau unitário

Como a transformada de Laplace da função degrau unitário é $\frac{1}{s}$.
Substituindo $R(s) = \frac{1}{s}$ na Equação 1.1, obtemos:

\begin{equation}
C(s) = \frac{1}{Ts + 1}\,\frac{1}{s}
\tag{1.2}
\end{equation}

Expandindo $C(s)$ em frações parciais, temos:

\begin{equation}
C(s) = \frac{1}{s} - \frac{T}{Ts + 1}
       = \frac{1}{s} - \frac{1}{s + (1/T)}
\tag{1.3}
\end{equation}

Considerando a transformada inversa de Laplace da Equação 1.3, obtemos:

\begin{equation}
c(t) = 1 - e^{-t/T}, \quad \text{para } t \ge 0
\tag{1.4}
\end{equation}

De acordo com a Equação 1.4, a resposta $c(t)$ é inicialmente zero e, no fim,
torna-se unitária.
Quanto menor a constante de tempo $T$, mais rapidamente o sistema
responde.

```python
%matplotlib inline
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

BASE_DIR = Path.cwd()
if not (BASE_DIR / "cssim.py").exists():
    BASE_DIR = BASE_DIR / "cssim"
if str(BASE_DIR) not in sys.path:
    sys.path.insert(0, str(BASE_DIR))

from cssim import load_json, run_simulation, set_parameters

# Define o parâmetro do sistema e as variáveis simbólicas.
T = 1.0
s = sp.symbols("s")
tempo = sp.symbols("t", nonnegative=True)

# Monta a função de transferência e a transformada do degrau unitário.
G_s = 1 / (T * s + 1)
R_s = 1 / s

# Calcula a saída no domínio de Laplace.
C_s = sp.cancel(G_s * R_s)

# Expande a saída em frações parciais.
C_s_fracoes = sp.apart(C_s, s)

# Aplica a transformada inversa de Laplace para obter c(t).
c_t = sp.inverse_laplace_transform(C_s_fracoes, s, tempo, noconds=True)
c_t = sp.simplify(c_t.subs(sp.Heaviside(tempo), 1))

# Converte a expressão simbólica em uma função numérica.
c_func = sp.lambdify(tempo, c_t, modules="numpy")

# Executa no CSSIM o mesmo caso usado na solução analítica.
params = {"T": T}
set_parameters(params)
data = load_json(str(BASE_DIR / "diagrams/first_order_step.json"))
t, _, _, outputs = run_simulation(
    data,
    step_size=float(data["step_size"]),
    simulation_time=float(data["sim_time"]),
    param_values=params,
 )
_, y_sim = next(iter(outputs.items()))
y_sim = np.asarray(y_sim, dtype=float)

# Avalia a solução analítica nos mesmos instantes da simulação.
y_theory = np.asarray(c_func(t), dtype=float)
max_abs_error = np.max(np.abs(y_sim - y_theory))

# Sobrepõe a curva analítica e a saída do CSSIM.
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(t, y_sim, label="CSSIM", linewidth=2)
ax.plot(t, y_theory, "--", label="Solução analítica", linewidth=2)
ax.set(
    xlabel="Tempo (s)",
    ylabel="Saída c(t)",
    title=f"Primeira ordem — degrau unitário | erro máximo = {max_abs_error:.3e}",
)
ax.grid(True)
ax.legend()
plt.show()
```

### Resposta à rampa unitária

Como a transformada de Laplace da rampa unitária é $\frac{1}{s^2}$, obtemos a saída do
sistema como:

\begin{equation}
C(s) = \frac{1}{Ts + 1} \frac{1}{s^2}
\tag{1.5}
\end{equation}

Expandindo $C(s)$ em frações parciais, temos:

\begin{equation}
C(s) = \frac{1}{s^2} - \frac{T}{s} + \frac{T^2}{Ts + 1}.
\tag{1.6}
\end{equation}

Aplicando a transformada inversa de Laplace da Equação 1.6, obtemos:

\begin{equation}
c(t) = t - T + T e^{-t/T}, \quad \text{para } t \ge 0.
\tag{1.7}
\end{equation}

O sinal de erro $e(t)$ é:

\begin{equation}
\begin{aligned}
e(t) &= r(t) - c(t) \\
     &= T \left(1 - e^{-t/T}\right).
\end{aligned}
\tag{1.8}
\end{equation}

Conforme $t$ tende ao infinito, $e^{-t/T}$ se aproxima de zero e, assim, o
sinal de erro $e(t)$ se aproxima de $T$ ou

\begin{equation}
e(\infty) = T
\tag{1.9}
\end{equation}
 
O erro do sistema para seguir a rampa unitária como sinal de entrada é igual a
$T$ para $t$ suficientemente grande. Quanto menor a constante de tempo $T$,
menor o erro estacionário ao seguir a entrada em rampa.

```python
%matplotlib inline
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

BASE_DIR = Path.cwd()
if not (BASE_DIR / "cssim.py").exists():
    BASE_DIR = BASE_DIR / "cssim"
if str(BASE_DIR) not in sys.path:
    sys.path.insert(0, str(BASE_DIR))

from cssim import load_json, run_simulation, set_parameters

# Define o parâmetro do sistema e as variáveis simbólicas.
T = 1.0
s = sp.symbols("s")
tempo = sp.symbols("t", nonnegative=True)

# Monta a função de transferência e a transformada da rampa unitária.
G_s = 1 / (T * s + 1)
R_s = 1 / s**2

# Calcula C(s) e realiza a expansão em frações parciais.
C_s = sp.cancel(G_s * R_s)
C_s_fracoes = sp.apart(C_s, s)

# Obtém a solução analítica c(t) pela transformada inversa de Laplace.
c_t = sp.inverse_laplace_transform(C_s_fracoes, s, tempo, noconds=True)
c_t = sp.simplify(c_t.subs(sp.Heaviside(tempo), 1))
c_func = sp.lambdify(tempo, c_t, modules="numpy")

# Executa no CSSIM o diagrama com entrada em rampa.
params = {"T": T}
set_parameters(params)
data = load_json(str(BASE_DIR / "diagrams/first_order_ramp.json"))
t, _, _, outputs = run_simulation(
    data,
    step_size=float(data["step_size"]),
    simulation_time=float(data["sim_time"]),
    param_values=params,
 )
_, y_sim = next(iter(outputs.items()))
y_sim = np.asarray(y_sim, dtype=float)

# Avalia a expressão analítica nos instantes produzidos pelo CSSIM.
y_theory = np.asarray(c_func(t), dtype=float)
max_abs_error = np.max(np.abs(y_sim - y_theory))

# Sobrepõe as duas respostas para realizar a validação.
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(t, y_sim, label="CSSIM", linewidth=2)
ax.plot(t, y_theory, "--", label="Solução analítica", linewidth=2)
ax.set(
    xlabel="Tempo (s)",
    ylabel="Saída c(t)",
    title=f"Primeira ordem — rampa unitária | erro máximo = {max_abs_error:.3e}",
)
ax.grid(True)
ax.legend()
plt.show()
```

## Sistemas de segunda ordem

<center>
<img src="img/second_order.png" alt="Sistema de segunda ordem" width="50%">
</center>

A seguir, analisaremos a resposta de um sistema de controle típico de segunda
ordem às entradas em degrau e rampa. Usaremos um servossistema como
exemplo de sistema de segunda ordem.

### Servossistema

A figura acima mostra um servossistema constituído por um controlador
proporcional e por elementos de carga (elementos de inércia e atrito viscoso).
O objetivo é controlar a posição de saída $c$ de acordo com a posição de
entrada $r$.

A equação de movimento para os elementos de carga pode ser escrita como

\begin{equation}
J \ddot{c}(t) + B \dot{c}(t) = T(t).
\tag{1.10}
\end{equation}

onde $T(t)$ é o torque aplicado pelo controlador proporcional, cujo ganho é
$K$. No domínio de Laplace, assumindo condições iniciais nulas, temos

\begin{equation}
Js^2 C(s) + Bs C(s) = T(s).
\tag{1.11}
\end{equation}

Logo, a função de transferência entre $C(s)$ e $T(s)$ é

\begin{equation}
\frac{C(s)}{T(s)} = \frac{1}{Js^2 + Bs}
= \frac{1}{s (Js + B)}.
\tag{1.12}
\end{equation}

O diagrama de blocos pode ser reorganizado para obter a função de transferência
de malha fechada entre $C(s)$ e $R(s)$:

\begin{equation}
\frac{C(s)}{R(s)} = \frac{K}{Js^2 + Bs + K}.
\tag{1.13}
\end{equation}

Esse sistema, cuja função de transferência de malha fechada possui dois polos,
é chamado de **sistema de segunda ordem**.

A função de transferência de malha fechada é dada por
\begin{equation}
\frac{C(s)}{R(s)}=
\frac{\frac{K}{J}}
{\left[s+\frac{B}{2J}+\sqrt{\left(\frac{B}{2J}\right)^{2}-\frac{K}{J}}\right]
 \left[s+\frac{B}{2J}-\sqrt{\left(\frac{B}{2J}\right)^{2}-\frac{K}{J}}\right]} .
\tag{1.14}
\end{equation}

Os polos de malha fechada são complexos conjugados se $B^{2}-4JK<0$ e são reais se $B^{2}-4JK\ge 0$.
Na análise da resposta transitória, é conveniente escrever:
\begin{equation}
\frac{K}{J}=\omega_{n}^{2},
\qquad
\frac{B}{J}=2\zeta\omega_{n}=2\sigma .
\tag{1.15}
\end{equation}

onde $\sigma$ é chamada *atenuação*; $\omega_{n}$ é a frequência natural não amortecida; e $\zeta$ é o coeficiente de amortecimento do sistema.
O coeficiente de amortecimento $\zeta$ é a relação entre o amortecimento real $B$ e o amortecimento crítico, ou
$B_{c}=2\sqrt{JK}$, ou ainda
\begin{equation}
\zeta=\frac{B}{B_{c}}=\frac{B}{2\sqrt{JK}} .
\tag{1.16}
\end{equation}

Em termos de $\zeta$ e $\omega_{n}$,a função de transferência de malha fechada $C(s)/R(s)$ pode ser escrita como:
\begin{equation}
\frac{C(s)}{R(s)}=\frac{\omega_{n}^{2}}{s^{2}+2\zeta\omega_{n}s+\omega_{n}^{2}} .
\tag{1.17}
\end{equation}

Essa forma é chamada *forma-padrão* do sistema de segunda ordem.

O comportamento dinâmico do sistema de segunda ordem pode ser descrito em termos de dois parâmetros $\zeta$ e $\omega_{n}$.
Se $0<\zeta<1$, os polos de malha fechada são complexos conjugados e se situam no semiplano esquerdo do plano $s$.
O sistema é então chamado subamortecido, e a resposta transitória é oscilatória.
Se $\zeta=0$, a resposta transitória não decai.
Se $\zeta=1$, o sistema é denominado criticamente amortecido.
Os sistemas superamortecidos correspondem a $\zeta>1$.



### Resposta ao degrau



```python
%matplotlib inline
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

BASE_DIR = Path.cwd()
if not (BASE_DIR / "cssim.py").exists():
    BASE_DIR = BASE_DIR / "cssim"
if str(BASE_DIR) not in sys.path:
    sys.path.insert(0, str(BASE_DIR))

from cssim import load_json, run_simulation, set_parameters

# Define os parâmetros e as variáveis simbólicas.
K = 1.0
J = 1.0
B = 1.0
s = sp.symbols("s")
tempo = sp.symbols("t", nonnegative=True)

# Monta a função de transferência de malha fechada.
G_mf_s = K / (J * s**2 + B * s + K)

# Multiplica pela transformada do degrau unitário.
R_s = 1 / s
C_s = sp.cancel(G_mf_s * R_s)

# Expande C(s) e calcula sua transformada inversa de Laplace.
C_s_fracoes = sp.apart(C_s, s)
c_t = sp.inverse_laplace_transform(C_s_fracoes, s, tempo, noconds=True)
c_t = sp.simplify(c_t.subs(sp.Heaviside(tempo), 1))
c_func = sp.lambdify(tempo, c_t, modules="numpy")

# Executa o caso correspondente no CSSIM.
params = {"K": K, "J": J, "B": B}
set_parameters(params)
data = load_json(str(BASE_DIR / "diagrams/second_order.json"))
t, _, _, outputs = run_simulation(
    data,
    step_size=float(data["step_size"]),
    simulation_time=float(data["sim_time"]),
    param_values=params,
 )
_, y_sim = next(iter(outputs.items()))
y_sim = np.asarray(y_sim, dtype=float)

# Avalia a solução analítica no vetor de tempo da simulação.
y_theory = np.asarray(c_func(t), dtype=float)
max_abs_error = np.max(np.abs(y_sim - y_theory))

# Sobrepõe a resposta analítica e a resposta simulada.
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(t, y_sim, label="CSSIM", linewidth=2)
ax.plot(t, y_theory, "--", label="Solução analítica", linewidth=2)
ax.set(
    xlabel="Tempo (s)",
    ylabel="Saída c(t)",
    title=f"Segunda ordem — degrau unitário | erro máximo = {max_abs_error:.3e}",
)
ax.grid(True)
ax.legend()
plt.show()
```

## Controladores automáticos

Um controlador automático ajusta o comportamento de um sistema comparando o valor real de saída da planta com a entrada de referência desejada. Com base nessa comparação, o controlador gera um sinal de controle que atua sobre a planta, minimizando o erro entre a saída e a referência.

### Controlador Proporcional

O controlador proporcional (P) ajusta a saída do sistema de acordo com a magnitude do erro entre a saída desejada e a saída real. A ação de controle é proporcional ao erro. Sendo a saída do controlador $u(t)$ e o sinal de erro atuante $e(t)$, a relação matemática do controlador proporcional é dada por:
\begin{equation}
  u(t) = K_p e(t)
\tag{1.18}
\end{equation}
onde $K_p$ é o ganho proporcional. Aplicando Laplace, a função de transferência do controlador proporcional é dada por:
\begin{equation}
  \frac{U(s)}{E(s)} = K_p.
\tag{1.19}
\end{equation}

<center>
<img src="img/controle_proporcional.png" alt="Sistema de controle proporcional" width="50%">
</center>

O controle de uma planta apenas proporcional pode não ser suficiente para atender aos requisitos de desempenho desejados. Em uma resposta ao degrau, por exemplo, o controle proporcional sem um integrador ocasiona em um erro em regime estacionário. Considerando o sistema da figura e definindo

\begin{equation}
  G(s) = \frac{K}{Ts + 1},
\tag{1.20}
\end{equation}

a função de transferência em malha fechada é dada por:

\begin{equation}
  \frac{C(s)}{R(s)} = \frac{G(s)}{1 + G(s)}.
\tag{1.21}
\end{equation}

O erro é definido como:
\begin{equation}
  E(s) = R(s) - C(s).
\tag{1.22}
\end{equation}

Dividindo ambos os lados da equação por $R(s)$, obtemos:

\begin{equation}
  \frac{E(s)}{R(s)} = 1 - \frac{C(s)}{R(s)}.
\tag{1.23}
\end{equation}

Substituindo a função de transferência em malha fechada, temos:

\begin{equation}
  \frac{E(s)}{R(s)} = 1 - \frac{G(s)}{1 + G(s)} = \frac{1}{1 + G(s)}.
\tag{1.24}
\end{equation}

Assim, o erro $E(s)$ pode ser expresso como:

\begin{equation}
  E(s) = \frac{1}{1 + G(s)} R(s) = \frac{1}{1 + \frac{K}{Ts + 1}} R(s) = \frac{Ts + 1}{Ts + 1 + K} R(s).
\tag{1.25}
\end{equation}

Para uma entrada degrau unitário, temos $R(s) = \frac{1}{s}$. Portanto, o erro é dado por:

\begin{equation}
  E(s) = \frac{Ts + 1}{Ts + 1 + K} \frac{1}{s}.
\tag{1.26}
\end{equation}

Aplicando o Teorema do Valor Final, o erro em regime estacionário é calculado como:

\begin{equation}
  e_{ss} = \lim_{t \to \infty} e(t) = \lim_{s \to 0} s E(s) = \lim_{s \to 0} s \cdot \frac{Ts + 1}{Ts + 1 + K} \cdot \frac{1}{s} = \frac{1}{1 + K}.  
\tag{1.27}
\end{equation}

Assim, o erro em regime estacionário é diferente de zero e depende do ganho proporcional $K$. Esse erro é chamado de erro residual. Para eliminá-lo, é necessário adicionar uma ação de controle integral ao sistema.

Execute a simulação abaixo para visualizar a ação de controle proporcional e observe o erro residual.

```python
%matplotlib inline
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

BASE_DIR = Path.cwd()
if not (BASE_DIR / "cssim.py").exists():
    BASE_DIR = BASE_DIR / "cssim"
if str(BASE_DIR) not in sys.path:
    sys.path.insert(0, str(BASE_DIR))

from cssim import load_json, run_simulation, set_parameters

# Define os parâmetros e as variáveis simbólicas.
T = 1.0
Kp = 2.0
s = sp.symbols("s")
tempo = sp.symbols("t", nonnegative=True)

# Monta o controlador P e a planta.
G_c_s = Kp
G_p_s = 1 / (T * s + 1)

# Fecha a malha com realimentação unitária.
G_mf_s = sp.cancel(G_c_s * G_p_s / (1 + G_c_s * G_p_s))

# Aplica o degrau, expande C(s) e calcula c(t).
R_s = 1 / s
C_s = sp.cancel(G_mf_s * R_s)
C_s_fracoes = sp.apart(C_s, s)
c_t = sp.inverse_laplace_transform(C_s_fracoes, s, tempo, noconds=True)
c_t = sp.simplify(c_t.subs(sp.Heaviside(tempo), 1))
c_func = sp.lambdify(tempo, c_t, modules="numpy")

# Executa o controlador proporcional no CSSIM.
params = {"T": T, "Kp": Kp}
set_parameters(params)
data = load_json(str(BASE_DIR / "diagrams/proportional.json"))
t, _, _, outputs = run_simulation(
    data,
    step_size=float(data["step_size"]),
    simulation_time=float(data["sim_time"]),
    param_values=params,
 )
_, y_sim = next(iter(outputs.items()))
y_sim = np.asarray(y_sim, dtype=float)

# Avalia a resposta analítica e calcula o erro da validação.
y_theory = np.asarray(c_func(t), dtype=float)
max_abs_error = np.max(np.abs(y_sim - y_theory))

# Compara visualmente a solução analítica com o CSSIM.
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(t, y_sim, label="CSSIM", linewidth=2)
ax.plot(t, y_theory, "--", label="Solução analítica", linewidth=2)
ax.set(
    xlabel="Tempo (s)",
    ylabel="Saída c(t)",
    title=f"Controlador P | erro máximo = {max_abs_error:.3e}",
)
ax.grid(True)
ax.legend()
plt.show()
```

### Controlador Integral



O controlador integral (I) ajusta o valor de saída do sistema $u(t)$ conforme uma taxa de variação proporcional ao sinal de erro atuante $e(t)$. A equação matemática do controlador integral é dada por:
\begin{equation}
  u(t) = K_i \int_{0}^{t} e(t) dt
\tag{1.28}
\end{equation}
onde $K_i$ é o ganho integral. A função de transferência do controlador integral é dada por:
\begin{equation}
  \frac{U(s)}{E(s)} = \frac{K_i}{s}.
\tag{1.29}
\end{equation}

A ação de controle integral é capaz de eliminar o erro em regime estacionário. O sinal de controle utilizado pelo controlador integral é a área sob a curva do sinal de erro ao longo do tempo.

<center>
<img src="img/controle_integral.png" alt="Sistema de controle integral" width="50%">
</center>

Considerando a figura acima, a função de transferência em malha fechada é dada por:

\begin{equation}
  \frac{C(s)}{R(s)} = \frac{\frac{K}{s(Ts + 1)}}{1 + \frac{K}{s(Ts + 1)}} = \frac{K}{s(Ts+1)+K}.
\tag{1.30}
\end{equation}

\begin{equation}
  E(s) = R(s) - C(s)
\tag{1.31}
\end{equation}

\begin{equation}
  \frac{E(s)}{R(s)} = 1 - \frac{C(s)}{R(s)} = 1 - \frac{K}{s(Ts+1)+K} = \frac{s(Ts+1)}{s(Ts+1)+K}.
\tag{1.32}
\end{equation}

Aplicando o Teorema do Valor Final para uma entrada degrau unitário, temos:

\begin{equation}
  e_{ss} = \lim_{t \to \infty} e(t) = \lim_{s \to 0} s E(s) = \lim_{s \to 0} s \frac{s(Ts+1)}{s(Ts+1)+K} \frac{1}{s} = 0.
\tag{1.33}
\end{equation}

Assim, vemos que o erro em regime estacionário é zero, ou seja, a ação de controle integral é capaz de eliminar o erro residual presente no sistema proporcional. No entanto, a ação de controle integral pode introduzir um comportamento oscilatório e instável no sistema, especialmente se o ganho integral for muito alto. Portanto, é importante ajustar cuidadosamente os parâmetros do controlador integral para garantir a estabilidade do sistema ou considerar a adição de um termo derivativo para melhorar a resposta transitória, como explicado a seguir.

Execute a simulação abaixo e observe a ação de controle integral.


```python
%matplotlib inline
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

BASE_DIR = Path.cwd()
if not (BASE_DIR / "cssim.py").exists():
    BASE_DIR = BASE_DIR / "cssim"
if str(BASE_DIR) not in sys.path:
    sys.path.insert(0, str(BASE_DIR))

from cssim import load_json, run_simulation, set_parameters

# Define os parâmetros e as variáveis simbólicas.
T = 1.0
Kp = 1.0
Ti = 1.0
s = sp.symbols("s")
tempo = sp.symbols("t", nonnegative=True)

# Monta o controlador PI implementado pelo bloco PID e a planta.
G_c_s = Kp * (1 + 1 / (Ti * s))
G_p_s = 1 / (T * s + 1)

# Calcula a função de transferência de malha fechada.
G_mf_s = sp.cancel(G_c_s * G_p_s / (1 + G_c_s * G_p_s))

# Aplica o degrau unitário e obtém a expansão em frações parciais.
R_s = 1 / s
C_s = sp.cancel(G_mf_s * R_s)
C_s_fracoes = sp.apart(C_s, s)

# Calcula a resposta analítica no domínio do tempo.
c_t = sp.inverse_laplace_transform(C_s_fracoes, s, tempo, noconds=True)
c_t = sp.simplify(c_t.subs(sp.Heaviside(tempo), 1))
c_func = sp.lambdify(tempo, c_t, modules="numpy")

# Executa no CSSIM o mesmo controlador PI.
params = {"T": T, "Kp": Kp, "Ti": Ti}
set_parameters(params)
data = load_json(str(BASE_DIR / "diagrams/integral.json"))
t, _, _, outputs = run_simulation(
    data,
    step_size=float(data["step_size"]),
    simulation_time=float(data["sim_time"]),
    param_values=params,
 )
_, y_sim = next(iter(outputs.items()))
y_sim = np.asarray(y_sim, dtype=float)

# Avalia a solução analítica e calcula o maior erro absoluto.
y_theory = np.asarray(c_func(t), dtype=float)
max_abs_error = np.max(np.abs(y_sim - y_theory))

# Sobrepõe as curvas para validar a implementação.
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(t, y_sim, label="CSSIM", linewidth=2)
ax.plot(t, y_theory, "--", label="Solução analítica", linewidth=2)
ax.set(
    xlabel="Tempo (s)",
    ylabel="Saída c(t)",
    title=f"Controlador PI | erro máximo = {max_abs_error:.3e}",
)
ax.grid(True)
ax.legend()
plt.show()
```

### Controlador Proporcional-Derivativo

O controlador proporcional-derivativo (PD) combina as ações dos controladores proporcional e derivativo. A equação matemática do controlador PD é dada por:
\begin{equation}
  u(t) = K_p e(t) + K_p T_d \frac{de(t)}{dt}
\tag{1.34}
\end{equation}
e a função de transferência do controlador PD é dada por:
\begin{equation}
  \frac{U(s)}{E(s)} = K_p (1 + T_d s).
\tag{1.35}
\end{equation}

A ação de controle derivativo antecipa o comportamento futuro do erro, respondendo à sua taxa de variação. Isso ajuda a reduzir o sobressinal e melhora a estabilidade do sistema, tornando a resposta mais rápida e menos oscilatória. O controle derivativo não afeta diretamente o erro em regime estacionário, mas pode melhorar significativamente a resposta transitória do sistema.

<center>
<img src="img/controle_proporcional_derivativo.png" alt="Sistema de controle proporcional-derivativo" width="60%">
</center>

Considerando a figura acima, a função de transferência em malha fechada é dada por:
\begin{equation}
  \frac{C(s)}{R(s)} = \frac{K_p(1 + T_d s)}{J s^2 + K_p T_d s + K_p}.
\tag{1.36}
\end{equation}

A equação característica do sistema é dada por:
\begin{equation}
  J s^2 + K_p T_d s + K_p = 0.
\tag{1.37}
\end{equation}

A adição do termo derivativo introduz um efeito de amortecimento na resposta do sistema, reduzindo o sobressinal e melhorando a estabilidade. Isso permite o uso de um ganho proporcional $K_p$ mais alto sem comprometer a estabilidade do sistema.

Execute a simulação abaixo para visualizar a ação de controle proporcional-derivativa. Experimente diferentes valores para Kp e Td e observe a resposta do sistema.


```python
%matplotlib inline
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import numpy as np
from scipy import signal

BASE_DIR = Path.cwd()
if not (BASE_DIR / "cssim.py").exists():
    BASE_DIR = BASE_DIR / "cssim"
if str(BASE_DIR) not in sys.path:
    sys.path.insert(0, str(BASE_DIR))

from cssim import load_json, run_simulation, set_parameters

# Define os parâmetros do controlador, da planta e do filtro derivativo.
Kp = 1.0
Td = 1.0
J = 1.0
N = 100.0
alpha = Td / N

# Escreve o controlador realizável como razão de polinômios.
# Gc(s) = Kp * [1 + Td*s/(alpha*s + 1)].
controller_num = np.array([Kp * (alpha + Td), Kp])
controller_den = np.array([alpha, 1.0])

# Escreve a planta Gp(s) = 1/(J*s²).
plant_num = np.array([1.0])
plant_den = np.array([J, 0.0, 0.0])

# Calcula numerador e denominador da malha fechada unitária.
open_num = np.polymul(controller_num, plant_num)
open_den = np.polymul(controller_den, plant_den)
closed_num = open_num
closed_den = np.polyadd(open_den, open_num)

# Inclui o polo da entrada degrau para formar C(s).
C_num = closed_num
C_den = np.polymul(closed_den, [1.0, 0.0])

# Realiza a expansão de C(s) em frações parciais.
residues, poles, direct_terms = signal.residue(C_num, C_den)
if direct_terms.size:
    raise ValueError("A solução analítica gerou um termo direto inesperado.")

# Executa no CSSIM o mesmo controlador PD filtrado.
params = {"Kp": Kp, "Td": Td, "J": J}
set_parameters(params)
data = load_json(str(BASE_DIR / "diagrams/derivative.json"))
t, _, _, outputs = run_simulation(
    data,
    step_size=float(data["step_size"]),
    simulation_time=float(data["sim_time"]),
    param_values=params,
 )
_, y_sim = next(iter(outputs.items()))
y_sim = np.asarray(y_sim, dtype=float)

# Aplica a transformada inversa: cada termo r/(s-p) vira r*exp(p*t).
y_theory_complex = np.zeros_like(t, dtype=complex)
for residue, pole in zip(residues, poles):
    y_theory_complex += residue * np.exp(pole * t)
y_theory = np.real_if_close(y_theory_complex).astype(float)
max_abs_error = np.max(np.abs(y_sim - y_theory))

# Sobrepõe a solução analítica e a simulação.
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(t, y_sim, label="CSSIM", linewidth=2)
ax.plot(t, y_theory, "--", label="Solução analítica", linewidth=2)
ax.set(
    xlabel="Tempo (s)",
    ylabel="Saída c(t)",
    title=f"Controlador PD com filtro derivativo | erro máximo = {max_abs_error:.3e}",
)
ax.grid(True)
ax.legend()
plt.show()
```

### Controladores PID

<center>
<img src="img/PID.png" alt="PID" width="50%">
</center>

A figura acima mostra o controle proporcional-integral-derivativo (PID) de uma planta.


Considerando o diagrama da imagem abaixo, deseja-se projetar um controlador PID capaz de rejeitar um distúrbio do tipo degrau unitário e, ao mesmo tempo, produzir uma resposta transitória satisfatória para a entrada de referência (2 a 3 s de tempo de acomodação usando um critério de 2%).

<center>
<img src="img/PID-disturbance.png" alt="PID-disturbance.png" width="60%">
</center>


A planta do sistema é descrita por:

\begin{equation}
\frac{1}{s^2 + 3{,}6s + 9}.
\tag{1.38}
\end{equation}


#### 1. Estrutura do controlador PID

O controlador 
PID é escrito na forma fatorada:

\begin{equation}
G_c(s)=\frac{K(as+1)(bs+1)}{s}.
\tag{1.39}
\end{equation}

Expandindo o numerador:

\begin{equation}
G_c(s)
=
\frac{K\left[abs^2+(a+b)s+1\right]}{s}.
\tag{1.40}
\end{equation}

Os parâmetros $K$, $a$ e $b$ serão escolhidos de modo que os polos de malha fechada atendam às especificações de desempenho.


#### 2. Função de transferência entre o distúrbio e a saída

Na ausência da entrada de referência, a função de transferência entre o distúrbio $D(s)$ e a saída $C_d(s)$ é

\begin{equation}
\frac{C_d(s)}{D(s)}
=
\frac{s}
{s(s^2+3{,}6s+9)+K(as+1)(bs+1)}.
\tag{1.41}
\end{equation}

Expandindo o denominador:

\begin{equation}
\frac{C_d(s)}{D(s)}
=
\frac{s}
{s^3+(3{,}6+Kab)s^2+(9+Ka+Kb)s+K}.
\tag{1.42}
\end{equation}

Portanto, a equação característica do sistema em malha fechada é

\begin{equation}
s^3+(3{,}6+Kab)s^2+(9+Ka+Kb)s+K=0.
\tag{1.43}
\end{equation}


#### 3. Escolha dos polos desejados

A especificação exige que a resposta ao distúrbio apresente um tempo de acomodação entre $2$ e $3$ segundos e um amortecimento razoável.

O tempo de acomodação representa o intervalo necessário para que a resposta do sistema se aproxime suficientemente de seu valor final e permaneça dentro de uma faixa de tolerância.

Adotando o critério de acomodação de $2\%$, considera-se que a resposta atingiu o regime permanente quando passa a permanecer entre $98\%$ e $102\%$ de seu valor final.

Para o critério de acomodação de $2\%$, o tempo de acomodação pode ser aproximado por:
\begin{equation}
t_s \approx \frac{4}{\zeta\omega_n}.
\tag{1.44}
\end{equation}

Assim, escolhendo

\begin{equation}
\zeta = 0{,}5,
\qquad
\omega_n=4\ \text{rad/s},
\tag{1.45}
\end{equation}

o tempo de acomodação será
\begin{equation}
t_s \approx \frac{4}{0,5 \times 4} = 2 \text{s}.
\tag{1.46}
\end{equation}

Para um sistema de segunda ordem, o polinômio correspondente é

\begin{equation}
s^2+2\zeta\omega_n s+\omega_n^2.
\tag{1.47}
\end{equation}

Substituindo os valores escolhidos:

\begin{equation}
s^2+2(0{,}5)(4)s+4^2 = s^2+4s+16.
\tag{1.48}
\end{equation}

Podemos escolher um terceiro polo em $s=-10$ para que o efeito deste polo real na resposta seja pequeno.

Assim, a equação característica desejada é

\begin{equation}
(s+10)(s^2+4s+16).
\tag{1.49}
\end{equation}

Expandindo:

\begin{equation}
(s+10)(s^2+4s+16)
=
s^3+14s^2+56s+160.
\tag{1.50}
\end{equation}


#### 4. Cálculo dos parâmetros do controlador

O polinômio característico obtido anteriormente foi

\begin{equation}
s^3+(3{,}6+Kab)s^2+(9+Ka+Kb)s+K.
\tag{1.51}
\end{equation}

Comparando seus coeficientes com os coeficientes do polinômio desejado,

\begin{equation}
s^3+14s^2+56s+160,
\tag{1.52}
\end{equation}

obtemos o seguinte sistema de equações:

\begin{equation}
\begin{aligned}
3{,}6+Kab &= 14, \\
9+Ka+Kb &= 56, \\
K &= 160.
\end{aligned}
\tag{1.53}
\end{equation}

Como $K=160$, segue que

\begin{equation}
ab
=
\frac{14-3{,}6}{160}
=
0{,}065,
\tag{1.54}
\end{equation}

e

\begin{equation}
a+b
=
\frac{56-9}{160}
=
0{,}29375.
\tag{1.55}
\end{equation}

Logo, o controlador PID pode ser escrito como

\begin{equation}
G_c(s)
=
\frac{K\left[abs^2+(a+b)s+1\right]}{s}.
\tag{1.56}
\end{equation}

Substituindo os valores encontrados:

\begin{equation}
G_c(s)
=
\frac{160\left(0{,}065s^2+0{,}29375s+1\right)}{s}.
\tag{1.57}
\end{equation}

Também é possível escrever o controlador na forma

\begin{equation}
G_c(s)
=
\frac{10{,}4\left(s^2+4{,}5192s+15{,}385\right)}{s}.
\tag{1.58}
\end{equation}


#### 5. Obtenção dos parâmetros $K_p$, $T_i$ e $T_d$

O controlador projetado é

\begin{equation}
G_c(s)=10{,}4s+47+\frac{160}{s}.
\tag{1.59}
\end{equation}

A forma padrão de um controlador PID é

\begin{equation}
G_c(s)=K_p \left(1+\frac{1}{T_i s}+T_d s \right).
\tag{1.60}
\end{equation}

Expandindo:

\begin{equation}
G_c(s) = K_p+\frac{K_p}{T_i s}+K_pT_d s.
\tag{1.61}
\end{equation}

Comparando os termos das duas expressões:

\begin{equation}
K_p=47,
\tag{1.62}
\end{equation}

\begin{equation}
\frac{K_p}{T_i}=160,
\tag{1.63}
\end{equation}

\begin{equation}
K_pT_d=10{,}4.
\tag{1.64}
\end{equation}

Logo,

\begin{equation}
T_i = \frac{47}{160} = 0{,}29375\ \text{s},
\tag{1.65}
\end{equation}

e

\begin{equation}
T_d = \frac{10{,}4}{47} \approx 0{,}2213\ \text{s}.
\tag{1.66}
\end{equation}

Assim, os parâmetros do controlador PID são

\begin{equation}
\boxed{
K_p=47,
\qquad
T_i=0{,}29375\ \text{s},
\qquad
T_d\approx0{,}2213\ \text{s}.
}
\tag{1.67}
\end{equation}


#### 6. Resposta ao distúrbio

Com o controlador projetado, a função de transferência entre o distúrbio e a saída é

\begin{equation}
C_d(s) = \frac{s} {s^3+14s^2+56s+160} D(s).
\tag{1.68}
\end{equation}

Usando a fatoração do denominador:

\begin{equation}
C_d(s) = \frac{s}{(s+10)(s^2+4s+16)} D(s).
\tag{1.69}
\end{equation}

Para um distúrbio do tipo degrau unitário,

\begin{equation}
D(s)=\frac{1}{s}.
\tag{1.70}
\end{equation}

Portanto,

\begin{equation}
C_d(s) = \frac{1}{(s+10)(s^2+4s+16)}.
\tag{1.71}
\end{equation}

Aplicando o Teorema do Valor Final:

\begin{equation}
\lim_{t\to\infty}c_d(t) = \lim_{s\to 0}sC_d(s).
\tag{1.72}
\end{equation}

Assim,

\begin{equation}
\lim_{t\to\infty}c_d(t) = \lim_{s\to 0} \frac{s}{(s+10)(s^2+4s+16)} = 0.
\tag{1.73}
\end{equation}

Isso mostra que o efeito permanente do distúrbio do tipo degrau é eliminado pelo controlador PID.


#### 7. Resposta à entrada de referência

Para a entrada de referência $R(s)$, a função de transferência de malha fechada é

\begin{equation}
\frac{C_r(s)}{R(s)} = \frac{10{,}4(s^2+4{,}5192s+15{,}385)}{s^3+14s^2+56s+160}.
\tag{1.74}
\end{equation}

Expandindo o numerador:

\begin{equation}
\frac{C_r(s)}{R(s)} = \frac{10{,}4s^2+47s+160}{s^3+14s^2+56s+160}.
\tag{1.75}
\end{equation}

Execute a simulação abaixo e observe as respostas à entrada e ao distúrbio.


```python
%matplotlib inline
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import numpy as np
from scipy import signal

BASE_DIR = Path.cwd()
if not (BASE_DIR / "cssim.py").exists():
    BASE_DIR = BASE_DIR / "cssim"
if str(BASE_DIR) not in sys.path:
    sys.path.insert(0, str(BASE_DIR))

from cssim import load_json, run_simulation, set_parameters

# Define os parâmetros projetados e o filtro derivativo do CSSIM.
Kp = 47.0
Ti = 0.29375
Td = 0.2213
N = 100.0
alpha = Td / N

# Escreve o PID realizável como razão de polinômios.
# Gc(s) = Kp * [1 + 1/(Ti*s) + Td*s/(alpha*s + 1)].
controller_num = np.array([
    Kp * Ti * (alpha + Td),
    Kp * (Ti + alpha),
    Kp,
 ])
controller_den = np.array([Ti * alpha, Ti, 0.0])

# Escreve a planta Gp(s) = 1/(s² + 3,6s + 9).
plant_num = np.array([1.0])
plant_den = np.array([1.0, 3.6, 9.0])

# Calcula o denominador comum da malha fechada.
open_num = np.polymul(controller_num, plant_num)
open_den = np.polymul(controller_den, plant_den)
closed_den = np.polyadd(open_den, open_num)

# Obtém os numeradores das respostas à referência e ao distúrbio.
reference_num = open_num
disturbance_num = controller_den

# Soma as duas respostas, pois referência e distúrbio são degraus unitários.
total_num = np.polyadd(reference_num, disturbance_num)
total_den = np.polymul(closed_den, [1.0, 0.0])

# Expande C(s) em frações parciais.
residues, poles, direct_terms = signal.residue(total_num, total_den)
if direct_terms.size:
    raise ValueError("A solução analítica gerou um termo direto inesperado.")

# Executa o diagrama com referência e distúrbio no CSSIM.
params = {"Kp": Kp, "Ti": Ti, "Td": Td}
set_parameters(params)
data = load_json(str(BASE_DIR / "diagrams/pid-disturbance.json"))
t, _, _, outputs = run_simulation(
    data,
    step_size=float(data["step_size"]),
    simulation_time=float(data["sim_time"]),
    param_values=params,
 )
_, y_sim = next(iter(outputs.items()))
y_sim = np.asarray(y_sim, dtype=float)

# Converte cada fração r/(s-p) no termo temporal r*exp(p*t).
y_theory_complex = np.zeros_like(t, dtype=complex)
for residue, pole in zip(residues, poles):
    y_theory_complex += residue * np.exp(pole * t)
y_theory = np.real_if_close(y_theory_complex).astype(float)
max_abs_error = np.max(np.abs(y_sim - y_theory))

# Sobrepõe a solução analítica e o resultado do CSSIM.
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(t, y_sim, label="CSSIM", linewidth=2)
ax.plot(t, y_theory, "--", label="Solução analítica", linewidth=2)
ax.set(
    xlabel="Tempo (s)",
    ylabel="Saída c(t)",
    title=f"PID — referência e distúrbio em degrau | erro máximo = {max_abs_error:.3e}",
)
ax.grid(True)
ax.legend()
plt.show()
```

### Circuito LRC
<center>
<img src="img/LRC_circuit.png" alt="Circuito LRC" width="30%%">
</center>

Nesta seção analisaremos o circuito LRC da imagem acima. Aplicando a lei das tensões de Kirchhoff ao sistema, obtemos as seguintes equações:

\begin{equation}
L\,\frac{di}{dt} + R\,i + \frac{1}{C}\int i\,dt = e_i
\tag{1.76}
\end{equation}

\begin{equation}
\frac{1}{C}\int i\,dt = e_o
\tag{1.77}
\end{equation}

Aplicando a transformada de Laplace e pondo condições iniciais nulas, obtemos:

\begin{equation}
L s I(s) + R I(s) + \frac{1}{C}\,\frac{1}{s}\,I(s) = E_i(s)
\tag{1.78}
\end{equation}

\begin{equation}
\frac{1}{C}\,\frac{1}{s}\,I(s) = E_o(s)
\tag{1.79}
\end{equation}

Admitindo que $e_i$ seja a entrada e $e_o$ a saída, a função de transferência do sistema será:

\begin{equation}
\frac{E_o(s)}{E_i(s)}=\frac{1}{LCs^{2}+RCs+1}
\tag{1.80}
\end{equation}

```python
%matplotlib inline
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

BASE_DIR = Path.cwd()
if not (BASE_DIR / "cssim.py").exists():
    BASE_DIR = BASE_DIR / "cssim"
if str(BASE_DIR) not in sys.path:
    sys.path.insert(0, str(BASE_DIR))

from cssim import load_json, run_simulation, set_parameters

# Define os componentes do circuito e as variáveis simbólicas.
L = 1.0
R = 20.0
C = 0.001
s = sp.symbols("s")
tempo = sp.symbols("t", nonnegative=True)

# Monta a função de transferência entre as tensões de saída e entrada.
G_s = 1 / (L * C * s**2 + R * C * s + 1)

# Aplica a transformada de Laplace do degrau unitário.
E_i_s = 1 / s
E_o_s = sp.cancel(G_s * E_i_s)

# Expande a saída em frações parciais.
E_o_s_fracoes = sp.apart(E_o_s, s)

# Calcula a tensão de saída analítica no domínio do tempo.
e_o_t = sp.inverse_laplace_transform(E_o_s_fracoes, s, tempo, noconds=True)
e_o_t = sp.simplify(e_o_t.subs(sp.Heaviside(tempo), 1))
e_o_func = sp.lambdify(tempo, e_o_t, modules="numpy")

# Executa o circuito LRC no CSSIM.
params = {"L": L, "R": R, "C": C}
set_parameters(params)
data = load_json(str(BASE_DIR / "diagrams/diagram_lrc.json"))
t, _, _, outputs = run_simulation(
    data,
    step_size=float(data["step_size"]),
    simulation_time=float(data["sim_time"]),
    param_values=params,
 )
_, y_sim = next(iter(outputs.items()))
y_sim = np.asarray(y_sim, dtype=float)

# Avalia a solução analítica nos mesmos instantes da simulação.
y_theory = np.asarray(e_o_func(t), dtype=float)
max_abs_error = np.max(np.abs(y_sim - y_theory))

# Sobrepõe as tensões obtidas analiticamente e pelo CSSIM.
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(t, y_sim, label="CSSIM", linewidth=2)
ax.plot(t, y_theory, "--", label="Solução analítica", linewidth=2)
ax.set(
    xlabel="Tempo (s)",
    ylabel="Tensão de saída eₒ(t)",
    title=f"Circuito LRC — degrau unitário | erro máximo = {max_abs_error:.3e}",
)
ax.grid(True)
ax.legend()
plt.show()
```
